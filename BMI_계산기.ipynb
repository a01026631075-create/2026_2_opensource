```python
def calculate_bmi():
    print("BMI (체질량지수) 계산기")

    try:
        height_cm = float(input("키(cm)를 입력하세요: "))
        weight_kg = float(input("몸무게(kg)를 입력하세요: "))

        height_m = height_cm / 100

        bmi = weight_kg / (height_m ** 2)

        print(f"\n당신의 BMI는 {bmi:.2f} 입니다.")

        if bmi < 18.5:
            print("결과: 저체중")
        elif 18.5 <= bmi < 23:
            print("결과: 정상")
        elif 23 <= bmi < 25:
            print("결과: 과체중")
        else:
            print("결과: 비만")

    except ValueError:
        print("키와 몸무게는 숫자로만 입력해 주세요.")

if __name__ == "__main__":
    calculate_bmi()
```
