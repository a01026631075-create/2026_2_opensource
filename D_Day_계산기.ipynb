```python
from datetime import datetime

def calculate_dday():
    print("D-Day 계산기")
    target_date_str = input("목표 날짜를 입력하세요 (예: 2026-12-25): ")

    try:
        target_date = datetime.strptime(target_date_str, "%Y-%m-%d").date()
        today = datetime.now().date()

        delta = target_date - today

        if delta.days > 0:
            print(f"⏳ 목표일까지 D-{delta.days} 남았습니다!")
        elif delta.days < 0:
            print(f"목표일로부터 D+{abs(delta.days)} 지났습니다!")
        else:
            print("바로 오늘입니다! D-Day!")

    except ValueError:
        print("날짜 형식을 YYYY-MM-DD에 맞게 정확히 입력해 주세요. (예: 2026-09-03)")

if __name__ == "__main__":
    calculate_dday()
```
