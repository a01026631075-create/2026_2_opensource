```python
import random

def play_rps():
    options = ['가위', '바위', '보']
    print("가위바위보 게임을 시작합니다! (종료하려면 '끝' 입력)")

    while True:
        user_choice = input("가위, 바위, 보 중 하나를 선택하세요: ")

        if user_choice == '끝':
            print("게임을 종료합니다.")
            break

        if user_choice not in options:
            print("올바른 입력이 아닙니다. '가위', '바위', '보' 중에서 입력해주세요.")
            continue

        comp_choice = random.choice(options)
        print(f"컴퓨터: {comp_choice}")

        # 승무패 로직 처리
        if user_choice == comp_choice:
            print("비겼습니다!")
        elif (user_choice == '가위' and comp_choice == '보') or \
             (user_choice == '바위' and comp_choice == '가위') or \
             (user_choice == '보' and comp_choice == '바위'):
            print("🎉 이겼습니다!")
        else:
            print("졌습니다.")

if __name__ == "__main__":
    play_rps()
```
