```python
def unit_converter():
    print("간단한 단위 변환기")
    print("1. 섭씨(℃) -> 화씨(℉)")
    print("2. cm -> inch")

    try:
        choice = input("원하는 변환의 번호를 입력하세요 (1 또는 2): ")

        if choice == '1':
            celsius = float(input("섭씨 온도를 입력하세요: "))
            fahrenheit = (celsius * 9/5) + 32
            print(f"{celsius}℃는 {fahrenheit:.2f}℉ 입니다.")

        elif choice == '2':
            cm = float(input("길이(cm)를 입력하세요: "))
            inch = cm / 2.54
            print(f"📏 {cm}cm는 {inch:.2f}inch 입니다.")

        else:
            print("1번이나 2번을 선택해 주세요.")

    except ValueError:
        print("문자 대신 올바른 숫자를 입력해 주세요.")

if __name__ == "__main__":
    unit_converter()
```
