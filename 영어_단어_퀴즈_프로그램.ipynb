```python
import random

def vocab_quiz():
    word_dict = {
        "commit": "커밋",
        "developer": "개발자",
        "variable": "변수",
        "function": "함수",
        "exception": "예외"
    }

    print("영어 단어 맞히기 퀴즈 (종료하려면 'q' 입력)")
    score = 0

    questions = list(word_dict.keys())
    random.shuffle(questions)

    for word in questions:
        try:
            answer = input(f"\n'{word}'의 뜻은 무엇일까요? : ").strip()

            if answer.lower() == 'q':
                print("\n퀴즈를 중도 종료합니다.")
                break

            if answer == word_dict[word]:
                print("정답입니다!")
                score += 1
            else:
                print(f"틀렸습니다. 정답은 '{word_dict[word]}' 입니다.")

        except KeyboardInterrupt:
            print("\n프로그램을 안전하게 종료합니다.")
            break

    print(f"\n최종 점수: {score} / {len(questions)}")

if __name__ == "__main__":
    vocab_quiz()
```
